```python
plt.figure(figsize=(7, 4))
plt.bar(values, pmf, width=0.15, edgecolor='black')
plt.xlabel("X = sqrt(pips)")
plt.ylabel("Empirical Probability")
plt.title("Empirical Mass Function of X (5000 rolls)")
plt.show()
```

```python
sorted_X = np.sort(X)
ecdf = np.arange(1, n+1) / n

plt.figure(figsize=(7, 4))
plt.step(sorted_X, ecdf, where='post')
plt.xlabel("X = sqrt(pips)")
plt.ylabel("ECDF")
plt.title("Empirical Cumulative Distribution Function of X")
plt.grid(True)
plt.show()
```

For the following distributions:

- Take a sample of 1000 draws $(x_1, x_2, ..., x_{1000})$ from the distribution, plot a KDE and ECDF, visually compare with the theoretical pdf/cdf

You can use https://docs.scipy.org/doc/scipy/reference/stats.html to generate values for the pdf/cdf and generate samples of random variates.


- Logistic distribution (similar to normal): 
$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$
with $\sigma >0$.
- Exponential distribution (similar to log-normal):
$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.
- Negative Binomial (similar to Poisson): The probability mass function for positive integers is:
$$
f(k;r,p) = \dfrac{(k+r-1)!}{k!(r-1)!}(1-p)^k p^r, \quad \text{ for $k=0,...n$}
$$
You can interpret this as follows: Flip a coin that comes up heads with probability $p$ until you get $r$ heads, and then stop. What is the probability of stopping at each $k=0,1,2,...$?
- Categorical (similar to Bernoulli): The probability mass function over $k = 1, 2, ..., K$ categories is
$$
f(k;p_1,...,p_K) = p_1^{k=1}p_2^{k=2}...p_K^{k=K}
$$
where $0 \le p_i \le 1$ and $ \sum_{k=1}^K p_k = 1$.

## Logistic Distribution

The support for the logistic distribution is all real numbers from -infinity to infinity because there is no value where e to the power of -x/theta is less than 0. 

The derivative is: 

![LogisticPDF](IMG_0393.jpg)


```python
sigmas = [1, 5, 0.5]
x = np.linspace(-10, 10, 500) 

for s in sigmas:
    pdf = logistic.pdf(x, loc=0, scale=s)
    plt.plot(x, pdf, label=f'σ={s}')
plt.title('Logistic Distribution PDF for different sigma')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
for s in sigmas:
    cdf = logistic.cdf(x, loc=0, scale=s)
    plt.plot(x, cdf, label=f'σ={s}')
plt.title('Logistic Distribution CDF for different sigmas')
plt.xlabel('x')
plt.ylabel('Cumulative Probability')
plt.legend()
plt.show()
```

```python
sigma = 2
rng = np.random.default_rng(100)
n = 1000
samp = logistic.rvs(loc=0, scale=sigma, size=n, random_state=rng)

sns.kdeplot(samp, label='Empirical KDE', fill=True)
plt.plot(x, logistic.pdf(x, loc=0, scale=sigma), color='C1', lw=2, label='Theoretical PDF')
plt.title(f'PDF of Logistic Distribution (σ={sigma})')
plt.xlabel('x')
plt.ylabel('Density')
plt.legend()
plt.show()
```

```python
x_sorted = np.sort(samp)
y_ecdf = np.arange(1, n+1) / n
plt.step(x_sorted, y_ecdf, where='post', label='Empirical ECDF')
plt.plot(x, logistic.cdf(x, loc=0, scale=sigma), color='C1', lw=2, label='Theoretical CDF')
plt.title(f'CDF of Logistic Distribution(σ={sigma})')
plt.xlabel('x')
plt.ylabel('Cumulative Probability')
plt.legend()
plt.show()
```

## Negative Binomial Distribution

the support for the negative binomial distribution is all discrete integers greater than or equal to 0.

```python
success = 10
p_values = [0.1, 0.2, 0.5, 0.9]  
x = np.arange(0, 50)  

plt.figure(figsize=(10, 6))
for p in p_values:
    pmf = nbinom.pmf(x, success, p)
    plt.scatter(x, pmf, marker='o', label=f'p={p}')

plt.xlabel('Number of failures (k)')
plt.ylabel('Probability')
plt.title(f'Negative Binomial PMF for various p values')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python

# Plot CDF for different p values
plt.figure(figsize=(10, 6))
for p in p_values:
    cdf = nbinom.cdf(x, success, p)
    plt.step(x, cdf, where='post', label=f'p={p}')

plt.xlabel('Number of failures (k)')
plt.ylabel('Cumulative Probability')
plt.title(f'Negative Binomial CDF for various p value')
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

```python
n = 10
p = 0.5
size = 1000
rng = np.random.default_rng(0)
sample = nbinom.rvs(n, p, size=size, random_state=rng)

x = np.arange(0, 30)
theoretical_pmf = nbinom.pmf(x, n, p)
sns.kdeplot(sample, label='Empirical')
plt.plot(x, theoretical_pmf, 'r-', label='Theoretical PMF')
plt.title('Negative Binomial PMF Comparison')
plt.xlabel('Number of failures')
plt.ylabel('Probability')
plt.legend()
plt.show()
```

```python
x_sorted = np.sort(sample)
y_ecdf = np.arange(1, len(sample) + 1) / len(sample)
theoretical_cdf = nbinom.cdf(x, n, p)
plt.step(x_sorted, y_ecdf, label='Empirical CDF')
plt.plot(x, theoretical_cdf, 'r-', label='Theoretical CDF')
plt.title('Negative Binomial CDF Comparison')
plt.xlabel('Number of failures')
plt.ylabel('Cumulative Probability')
plt.legend()
plt.show()
```

## Categorical Distribution

The support for the categorical distribution is discrete integer values from 1 to K.

```python
param_sets = {
    "Uniform": [0.25, 0.25, 0.25, 0.25],
    "Skewed": [0.6, 0.2, 0.1, 0.1],
    "Moderate": [0.4, 0.3, 0.2, 0.1],
}

categories = np.arange(1, 5)  

fig, axes = plt.subplots(1, len(param_sets), figsize=(15, 4))

for ax, (title, probs) in zip(axes, param_sets.items()):
    ax.bar(categories, probs, color="skyblue", edgecolor="black")
    ax.set_title(f"Categorical Distribution ({title})")
    ax.set_xlabel("Category")
    ax.set_ylabel("Probability")
    ax.set_xticks(categories)
    ax.set_ylim(0, 1)

plt.tight_layout()
plt.show()


fig, axes = plt.subplots(1, len(param_sets), figsize=(15, 4))

n_samples = 5000

for ax, (title, probs) in zip(axes, param_sets.items()):
    samples = np.random.choice(categories, size=n_samples, p=probs)

    sns.histplot(samples, bins=np.arange(0.5, 4.6, 1), ax=ax, stat="probability", color="lightcoral")
    ax.set_title(f"Sampled Distribution ({title})")
    ax.set_xlabel("Category")
    ax.set_ylabel("Empirical Probability")
    ax.set_xticks(categories)
    ax.set_ylim(0, 1)

plt.tight_layout()
plt.show()
```

```python
from statsmodels.distributions.empirical_distribution import ECDF

p = np.array([0.1, 0.2, 0.4, 0.2, 0.1])  # probabilities must sum to 1
categories = np.arange(1, len(p) + 1)

n = 1000
samples = np.random.choice(categories, size=n, p=p)

pmf = p
cdf = np.cumsum(p)

ecdf = ECDF(samples)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.kdeplot(samples, ax=axes[0], bw_adjust=0.5, fill=True, label='Empirical KDE')
axes[0].stem(categories, pmf, basefmt=" ", linefmt='r-', markerfmt='ro', label='Theoretical PMF')
axes[0].set_title('Categorical Distribution — KDE vs PMF')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Density')
axes[0].legend()

x_vals = np.linspace(1, len(p), 500)
axes[1].plot(x_vals, ecdf(x_vals), label='Empirical ECDF')
axes[1].step(categories, cdf, where='post', color='r', label='Theoretical CDF')
axes[1].set_title('Categorical Distribution — ECDF vs CDF')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Cumulative Probability')
axes[1].legend()

plt.tight_layout()
plt.show()

```
