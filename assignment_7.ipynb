## 1. 

- What is the expected value of a single die roll? 

Assuming this is a fair die, every outcome {1, 2, 3, 4, 5, 6} will have a probability of 1/6. Calculating this, we get: (1 + 2 + 3 + 4 + 5 + 6)/6 = 3.5. 
The expected value of a single die roll is 3.5.

- What is the expected value of rolling two dice and adding the results together?

If you add the two results together, you get 3.5 + 3.5 = 7. So the expected value of rolling two dice is 7.

- What is the expected winnings of any gamble in European roulette?

To play in European roulette you need to pay one dollar to bet on K slots. You can win with a probability of K/37 and gain 36/K dollars. You lose with (37-K)/37. As a result, the expected payoff is: 

Expected value: (K/37)(36/K) + ([37-K]/37)(0) - 1 = -0.027. 
So the expected value for a roulette bet gives an expected payoff of -0.03. 


- Imagine you roll a die, and you record the value you get. But, if you roll a six, you roll again, and add that value. What is the expected value?

So we're going to take the same expected value formula from the number 1-5 to get this: (1 + 2 + 3 + 4 + 5)/6
In the case that you roll a 6, you need to add this term to the expected value formula: (1/6)(6 + 1/6 + 2/6 + 3/6 + 4/6 + 5/6 + 6/6). 

So the entire formula would be: 

[(1 + 2 + 3 + 4 + 5)/6] + [(1/6)(6 + 1/6 + 2/6 + 3/6 + 4/6 + 5/6 + 6/6)] = 4.083. 

- Imagine that the process described in the last question continues until you fail to roll a six. What is the expected value of the process? (This can be tricky, you can simulate it to get an answer if you prefer. Hint: The answer is 4.2.)

J = roll again 

J = 1/6 + 2/6 + 3/6 + 4/6 + 5/6 + (6 + J)/6 
J = 21/6 + 1/6J 
J = 21/5 = 4.2 

```python
import numpy as np 

rng = np.random.default_rng(100) 

def roll_6(): 
    sum = 0 
    on = 1 
    while on == 1: 
        roll = rng.integers(1,7) 
        sum += roll 
        if( roll < 6 ): 
            on = 0 
    return sum

sample = 1000000
results = [roll_6() for i in range(sample)]

print("Mean: ", np.mean(results))
print("Standard Deviation: ", np.std(results))
```

Show, by example, that $v(\mathbb{E}[X]) \neq \mathbb{E}[v(X)]$, if $v(x) \neq a+bx$.

```python
import numpy as np 
import seaborn as sns 
import matplotlib.pyplot as plt 

p = 0.3
N = 100000 

X = np.random.binomial(1, p, size=N)

# v(x) = x^2
v = X**2

# v(E[X]) = (E[X])^2
v_E = (np.mean(X))**2

# E[v(X)] = E[X^2]
E_v = np.mean(v)

labels = ['E[X]^2', 'E[X^2]']
values = [v_E, E_v]

plt.figure(figsize=(10, 5))
bars = plt.bar(labels, values)

for bar, value in zip(bars, values):
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{value:.4f}",
        ha='center',
        va='bottom',
        fontsize=12
    )

plt.title("Comparison: v(E[X]) vs E[v(X)] for v(x)=x^2, X~Bernoulli(0.3)")
plt.ylabel("Value")
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```

These values are very different -- which shows that $v(\mathbb{E}[X]) \neq \mathbb{E}[v(X)]$, if $v(x) \neq a+bx$.

```python
df_9 = pd.read_csv(('./data/ames_prices.csv'))
y = np.log(df_9['price'])
sns.histplot(y, stat = 'proportion')
```

We took the log of this distribution because after tthe transformation, it is reasonably symmetric and bell-shaped around a mean of 12. Price is log-normally distributed. 

![LogisticPDF](IMG_0423.jpg)

```python
import scipy.stats as stats

mu_hat = np.mean(y)
sigma_hat = np.std(y, ddof=0) 


grid = np.linspace(9, 14, 100)


dist = stats.norm(loc=mu_hat, scale=sigma_hat)
pdf = dist.pdf(grid)
cdf = dist.cdf(grid)


fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.kdeplot(y, ax=axes[0], label="KDE")
sns.lineplot(x=grid, y=pdf, ax=axes[0], label="Normal PDF", linestyle="--")
axes[0].set_title("Density Comparison")
axes[0].legend()
sns.ecdfplot(y, ax=axes[1], label="ECDF")
sns.lineplot(x=grid, y=cdf, ax=axes[1], label="Normal CDF", linestyle="--")
axes[1].set_title("CDF Comparison")
axes[1].legend()

plt.tight_layout()
plt.show()
```

The fit of the ECDF is very good because the lines are basically overlapping each other. The data has a higher peak and greater leftward skew in comparison to the pdf. The general shape is very similar (bell-shaped) and the peak is very narrow. 

```python

def mle(y):
    mu_hat = np.mean(y)
    sigma_hat = np.std(y, ddof=0)
    return mu_hat, sigma_hat


S = 5000
bootstrap_estimates = np.array([mle(y.sample(frac=1.0, replace=True)) 
                                for _ in range(S)])

mu_hat_samples = bootstrap_estimates[:, 0]
sigma_hat_samples = bootstrap_estimates[:, 1]


def plot_sampling_distribution(samples, label):
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))

    sns.kdeplot(samples, ax=axes[0], label="KDE")
    axes[0].set_title(f"Sampling Density, ${label}$")
    axes[0].legend()

    sns.ecdfplot(samples, ax=axes[1], label="ECDF")
    axes[1].set_title(f"Sampling Distribution, ${label}$")
    axes[1].legend()

    plt.tight_layout()
    plt.show()

plot_sampling_distribution(mu_hat_samples, "\\hat{\\mu}")
plot_sampling_distribution(sigma_hat_samples, "\\hat{\\sigma}")
```
